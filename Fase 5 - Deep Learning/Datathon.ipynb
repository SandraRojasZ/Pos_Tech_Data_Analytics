```python
url = 'https://raw.githubusercontent.com/SandraRojasZ/Pos_Tech_Data_Analytics/main/Base_de_Dados/prospects/prospects.json'
```

```python
# prospects
with urlopen(url) as response:
    data1 = json.loads(response.read().decode())
```

**PROSPECTS**

```python
print(data1)
```

```python
# Passo 1: Criar uma lista para armazenar os dados dos prospects
dados_prospects = []
# Passo 2: Extrair código e dados dos prospects
for key, value in data1.items():
    codigo = key
    titulo = value['titulo']
    modalidade = value['modalidade']

    for prospect in value['prospects']:
        dados_prospects.append({
            'codigo': codigo,
            'titulo': titulo,
            'modalidade': modalidade,
            'nome': prospect.get('nome', ''),
            'codigo_prospect': prospect.get('codigo', ''),
            'situacao_candidato': prospect.get('situacao_candidado', ''),
            'data_candidatura': prospect.get('data_candidatura', ''),
            'ultima_atualizacao': prospect.get('ultima_atualizacao', ''),
            'comentario': prospect.get('comentario', ''),
            'recrutador': prospect.get('recrutador', '')
        })
# Passo 3: Criar um DataFrame
df_prospects = pd.DataFrame(dados_prospects)
```

```python
# Cria um dicionário para armazenar o número de strings vazias por coluna
strings_vazias = {}
# Itera sobre as colunas do DataFrame
for coluna in df_prospects.columns:
    # Conta strings vazias na coluna
    num_vazias = (df_prospects[coluna] == '').sum()
    strings_vazias[coluna] = num_vazias
# Mostra o resultado
for coluna, num_vazias in strings_vazias.items():
    print(f"'{coluna}': {num_vazias}")
```

```python
# Criar o DataFrame
applications = pd.DataFrame.from_dict(data2, orient='index')
# Expandir as colunas que contêm dicionários
infos_basicas_df = applications['infos_basicas'].apply(pd.Series)
informacoes_pessoais_df = applications['informacoes_pessoais'].apply(pd.Series)
informacoes_profissionais_df = applications['informacoes_profissionais'].apply(pd.Series)
formacao_e_idiomas_df = applications['formacao_e_idiomas'].apply(pd.Series)
# Concatenar os DataFrames expandidos com o DataFrame original
expanded_df = pd.concat([applications.drop(columns=['infos_basicas', 'informacoes_pessoais', 'informacoes_profissionais', 'formacao_e_idiomas']),
                         infos_basicas_df,
                         informacoes_pessoais_df,
                         informacoes_profissionais_df,
                         formacao_e_idiomas_df], axis=1)
```
